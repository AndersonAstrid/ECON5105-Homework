```python
# 1
import numpy as np
from scipy.integrate import quad

# define function f(x) = x^2
def integrand(x):
    return x**2

# integration bounds: x=0 to x=2
lower_bound = 0
upper_bound = 2

# run integration
result, error = quad(integrand, lower_bound, upper_bound)
print(f"Integral Result: {result}")
print(f"Estimated Error: {error}")
```

```python
# 2
#define the function f(x) = x^2
def f(x):
    return x**2

# estimate integral using rectangles
def estimated_area(a, b, n):
    # a = starating point, b = end point, n = number of rectangles

    width = (b - a) / n # width of each rectangle
    total_area = 0

    for i in range(n):
        # x-coordinate of left side of rectangle
        x = a + i * width

        # height of rectangle is f(x)
        height = f(x)

        # add area of this rectangle
        rectangle_area = width * height
        total_area = total_area + rectangle_area

    return total_area

actual_area = 8/3
estimated = estimated_area(0, 2, 10)

print("Actual area:", f"{actual_area:.6f}")
print("Estimated area:", f"{estimated:.6f}")
```
