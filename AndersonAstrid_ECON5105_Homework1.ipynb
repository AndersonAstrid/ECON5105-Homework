```python
import numpy as np
```

```python
# Question 1 a, b, c
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])
C = A + B
print("A + B:", C)
D = A - B
print("A - B:", D)
two_A = 2 * A
print("2A:", two_A)
AB = A @ B
print("AB:", AB)
BA = B @ A
print("BA:", BA)
```

```python
# Question 3
A = np.array([[2, 1], [3, -1]])
b = np.array([5, 4])
x = np.linalg.inv(A) @ b
print("Solutions [x, y]:", x)
```

```python
# Question 4 a & b
C = np.array([[1, 2, 3], [1, 1, 1], [3, 5, 7]])
determinant = np.linalg.det(C)
print("Determinant:", determinant)
rank = np.linalg.matrix_rank(C)
print("Rank C:", rank)
```

```python
# Question 5
D = np.array([[2, 1, 0], [1, 1, 1], [0, 1, 1]])
inverse = np.linalg.inv(D)
print("Inverse D:", inverse)
```
